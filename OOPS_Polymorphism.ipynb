# Unit 1 – Polymorphism (Method Based)
# Create two different classes that contain a method with the same name. Write a common function that accepts an object and calls that method to demonstrate polymorphism.
# Write another example where multiple objects are stored in a list and the same method is called using a loop to show how Python handles method calls dynamically
 

```python
class Dog:
    def sound (self):
        print("Dog Barks")

class Cat :
    def sound (self):
        print("Cat meow")

animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
```

#  Unit 2 – Polymorphism using Inheritance (Method Overriding)
#  Create a base class with a method. Then create one or more derived classes that override this method and show how different objects call their own version of the method.
# Also demonstrate this concept using multiple child classes stored in a list and executed using a loop.
 

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog Bark")


class Cat(Animal):
    def sound(self):
        print("Cat Meow")


class Cow(Animal):
    def sound(self):
        print("Cow Moo Moo")

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.sound()
```

# Unit 3 – Operator Overloading
# Create a class that represents a 3D object and includes a method to calculate its volume. Overload the + operator so that when two objects are added, the result gives the combined volume.


```python
class Box:
    def __init__(self,length,width,height):
        self.length = length 
        self.width = width 
        self.height = height 

    def volume(self):
        return self.length * self.width * self.height

    def __add__(self, other):
        return self.volume() + other.volume()

box1 = Box(2, 3, 4)
box2 = Box(5, 2, 3)

print("Volume of Object 1:", box1.volume())
print("Volume of Object 2:", box2.volume())

print("box Volume:", box1 + box2)
```

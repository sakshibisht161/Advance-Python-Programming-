# Question 7
# Create a class named Bank with a private variable __balance initialized as 2000. Create methods to:
# Deposit money into the account.
# Display the current balance using a Getter Method.


```python
class Bank:
    def __init__(self):
        self.__balance = 2000

    def deposit(self, amount):
        self.__balance = self.__balance + amount

    def get_balance(self):
        return self.__balance


bank = Bank()

bank.deposit(500)

print(bank.get_balance())
```

# Question 8
# Create a class named Exam with a private variable __marks. Write a Setter Method that accepts marks only between 0 and 100. If the entered marks are outside this range, display "Marks should be between 0 and 100"


```python
class Exam:
    def __init__(self):
        self.__marks = 0

    def set_marks(self, marks):
        if marks >= 0 and marks <= 100:
            self.__marks = marks
        else:
            print("Marks should be between 0 and 100")


exam = Exam()

exam.set_marks(85)

print(exam._Exam__marks)
```

# A public variable name

# Question 10
# Create a class named LibraryBook with private variables __title and __author. Create Setter Methods to update both values and Getter Methods to display them. Create an object, assign values using the setters, and print the details using the getters.


```python
class LibraryBook:
    def __init__(self):
        self.__title = ""
        self.__author = ""

    def set_title(self, title):
        self.__title = title

    def set_author(self, author):
        self.__author = author

    def get_title(self):
        return self.__title

    def get_author(self):
        return self.__author


book = LibraryBook()

book.set_title("Python Programming")
book.set_author("Guido van Rossum")

print(book.get_title())
print(book.get_author())
```

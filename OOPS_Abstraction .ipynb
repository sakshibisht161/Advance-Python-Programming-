# Abstraction 

```python
from abc import ABC
```

```python
class Car(ABC):
    def mileage (self):
        pass 
    def no_of_in_use_tyres(self):
        print("4")

class Suzuki(Car):
    def mileage (self):
        print("The mileage is 35kmph")

class Honda(Car):
    def mileage (self):
        print("The mileage is 24kmph")

s1 = Suzuki()
s1.mileage()
s1.no_of_in_use_tyres()
s2 = Honda()
s2.mileage()
s2.no_of_in_use_tyres()
```

# Create an abstract class with one abstract method and two subclasses that implement this method with different outputs

```python
from abc import ABC 
```

```python
class Animal(ABC):
    def make_sound(self):
        pass 

class Dog(Animal):
    def make_sound(self):
        print("Bark")

class Cat(Animal):
    def make_sound(self):
        print("Meow")

a1 = Dog()
a1.make_sound()
a2 = Cat()
a2.make_sound()
```

# Create an abstract class for vehicles with an abstract method for maximum speed and implement it in at least two subclasses 

```python
from abc import ABC
```

```python
class Vehicles(ABC):
    def maximum_speed(self):
        pass 

class Car(Vehicles):
    def maximum_speed(self):
        print("200km/h")

class Bike(Vehicles):
    def maximum_speed(self):
        print("150km/h")

car = Car()
car.maximum_speed()
bike = Bike()
bike.maximum_speed()
```

# Create an abstract class for shapes with an abstract method to calculate area and implement it in a subclass like circle

```python
from abc import ABC 
```

```python
class Shape(ABC):
    def calculate_area(self):
        pass 

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius 
        print("radius of circle:",self.radius)

    def area (self):
        return self.radius * self.radius

ac = Circle(5)
print("Area of circle:", ac.area())     
```

# Create an abstract class with a method for security and another normal method for database connection, then implement the abstract method in a subclass and demonstrate all methods

```python
from abc import ABC

class Database(ABC):

    def security(self):
        print("Security check required")

    def database_connection(self):
        print("Database connected successfully")


class MySQL(Database):

    def security(self):
        print("Security verified using password")

db = MySQL()
db.security()
db.database_connection()
```

# Question 1: Single Inheritance

# Write a Python program to create a class Vehicle with a method show_brand() that displays the vehicle brand. Create another class Car that inherits from Vehicle and has a method show_model(). Create an object of the Car class and display both the brand and model.


```python
class Vehical :
    def __init__(self , brand):
        self.brand = brand 

class car (Vehical):
    def __init__(self , brand , model):
        super().__init__(brand)
        self.model = model

    def display(self):
        self.brand = brand 
        self.model = model 

Car = car("maruti" , "suzuki")
print(Car.brand, Car.model)
        
```

# Question 2: Multiple Inheritance

# Write a Python program to create two parent classes Sports and Music. The Sports class should contain a method play_sport(), and the Music class should contain a method play_music(). Create a child class Student that inherits from both parent classes and has a method study(). Create an object of the Student class and call all three methods.


```python
class Sports:
    def play_sport(self):
        print("I play football.")


class Music:
    def play_music(self):
        print("I play guitar.")


class Student(Sports, Music):
    def study(self):
        print("I study Data Science.")

student = Student()

student.play_sport()
student.play_music()
student.study()
```

# Question 3: Multilevel Inheritance


# Write a Python program to create three classes: Vehicle, Car, and ElectricCar. The Car class should inherit from Vehicle, and the ElectricCar class should inherit from Car. Create methods to display the vehicle brand, car model, and battery capacity. Create an object of the ElectricCar class and call all the methods.


```python
class Vehical :
    def __init__(self,brand):
        self.brand = brand 
        
class Car(Vehical):
    def __init__(self,brand, model):
        super().__init__(brand)
        self.model = model 

class Electric_Car(Car):
    def __init__(self,brand , model ,battery):
        super().__init__(brand , model)
        self.battery = battery

    def display(self):
        self.brand = brand 
        self.model = model 
        self.battery = battery

car = Electric_Car("Tata","Punch Ev" , "35 kWh")
print(car.brand , car.model , car.battery)
    
    
```

# Question 4: Hierarchical Inheritance

# Write a Python program to create a parent class Shape with a method draw(). Create two child classes Circle and Rectangle that inherit from Shape. Add a method in each child class to display the shape name. Create objects of both child classes and call all the methods.


```python
class Shape:
    def __init__(self, name):
        self.name = name

    def draw(self):
        print("Drawing Shape")


class Circle(Shape):
    def __init__(self, name):
        super().__init__(name)

    def display(self):
        print(self.name)


class Rectangle(Shape):
    def __init__(self, name):
        super().__init__(name)

    def display(self):
        print(self.name)


circle = Circle("Circle")
rectangle = Rectangle("Rectangle")

circle.draw()
circle.display()
rectangle.draw()
rectangle.display()
```

# Question 5: Hybrid Inheritance

# Write a Python program to demonstrate Hybrid Inheritance using the following classes:
# Device (Parent Class)
# Laptop (inherits from Device)
# Tablet (inherits from Device)
# HybridDevice (inherits from both Laptop and Tablet)
 

```python
class Device:
    def __init__(self, brand):
        self.brand = brand

class Laptop(Device):
    def __init__(self, brand, laptop_model):
        super().__init__(brand)
        self.laptop_model = laptop_model

class Tablet(Device):
    def __init__(self, brand, tablet_model):
        super().__init__(brand)
        self.tablet_model = tablet_model

class Hybrid_Device(Laptop, Tablet):
    def __init__(self, brand, laptop_model, tablet_model):
        Device.__init__(self, brand)
        self.laptop_model = laptop_model
        self.tablet_model = tablet_model

    def display(self):
        print(self.brand, self.laptop_model, self.tablet_model)


device = Hybrid_Device("Dell", "Latitude", "Dell Tablet")
device.display()
```

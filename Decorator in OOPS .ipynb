# A Decorator in Python is a FUNCTIONS WHICH TAKES OTHER FUNCTIONS AS INPUT, ADD ADDITIONAL FUNCTIONALITIES AND RETURNS IT.
# FUNCTION--DECORATOR--FUNCTION bold text -IT IS A CALLABLE PYTHON OBJECT WHICH MODIFIES OTHER FUNCTION/CLASS.
# Think of a decorator as a wrapper: it takes a function, adds some kind of functionality, and then returns a new function with the added behavior.

# With Decorator

```python
def Decor(func):
    def wrapper():
        print("not start")
        func()
    return wrapper
@Decor
def printer():
    print("Hello1")
    print("Hello2")
printer()
```

```python
def Decor(func):
    def inner():
        print("Python1")
        func()
    return inner 
@Decor 
def printer():
    print("Python2")
    print("Pyhton3")
printer()
```

# First define the decorator function

```python
def Decorator_function(func):
    def wrapper():
        print("Before the original function")
        func()
        print("After the original function")
    return wrapper
@Decorator_function
def original_function():
    print("Inside original Function")
    pass
original_function()
```

# Decorators with Arguments
# What if your function takes arguments? Modify the wrapper to accept *args and **kwargs. 

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before call")
        result = func(*args, **kwargs)
        print("After call")
        return result
    return wrapper
@my_decorator
def say_hello(name):
    print(f"Hello, {name}!")
say_hello("Alice")
```

# 1. Logging Decorator (Fixed Arguments)

# Question: Create a decorator that logs function name and parameters. The function should take exactly 2 arguments.

```python
def log_decorator(func):
    def wrapper(a, b):
        print(f"Calling {func.__name__} with arguments: {a}, {b}")
        return func(a, b)
    return wrapper
@log_decorator
def multiply(x, y):
    return x * y
print(multiply(4, 5))
```

# Question : Create a decorator that logs when a function starts and when it finishes. The function should take exactly 2 arguments.

```python
def log_function(func):
    def wrapper(a, b):
        print("Function started")
        result = func(a, b)
        print("Function finished")
        return result
    return wrapper
@log_function
def add(a, b):
    return a + b
print(add(10, 20))
```

# Question : Create a decorator that prints the name of the function and the values of its two parameters.

```python
def logger(func):
    def wrapper(a, b):
        print("Function:", func.__name__)
        print("Parameters:", a, b)
        return func(a, b)
    return wrapper
@logger
def divide(a, b):
    return a / b
print(divide(20, 5))
```

# Timer Decorator (No Arguments)
# Question: Create a decorator that measures execution time of a function with no parameters.

```python
import time 
def timer_decorator(func):
    def wrapper():
        start = time.time()
        result = func()
        end = time.time()
        print(f"Time takem : {end - start :.4f} seconds")
        return result
    return wrapper 
@timer_decorator 
def do_work():
    time.sleep(1)
    print("work_done")
do_work()
```

# Question : Create a timer decorator that prints "Function started" before execution and displays the total execution time after the function finishes.

```python
import time

def timer(func):
    def wrapper(a, b):
        print("Function started")
        start = time.time()
        result = func(a, b)
        end = time.time()
        print("Execution Time:", end - start, "seconds")
        return result
    return wrapper
@timer
def multiply(a, b):
    return a * b
print(multiply(5, 4))
```

# Question : Create a timer decorator to calculate how much time a function takes to execute. Apply it to a function calculate(a, b) that returns the product of two numbers.

```python
import time

def timer(func):
    def wrapper(a, b):
        start = time.time()
        result = func(a, b)
        end = time.time()
        print("Time taken:", end - start, "seconds")
        return result
    return wrapper
@timer
def calculate(a, b):
    return a * b
print(calculate(10, 5))
```

# 3. Authorization Decorator (Single String Parameter)

# Question: Create a decorator that checks if a user is "admin".

```python
def requires_admin(func):
    def wrapper(user):
        if user != "admin":
            print("Access denied")
            return
        return func(user)
    return wrapper
@requires_admin
def access_dashboard(user):
    print(f"{user} accessed the dashboard.")
access_dashboard("guest")
access_dashboard("admin")
```

# Question Create an authorization decorator that accepts a single string parameter representing the user's role. Allow access only for the "manager" role.

```python
def authorization(role):
    def decorator(func):
        def wrapper():
            if role == "manager":
                func()
            else:
                print("Access Denied")
        return wrapper
    return decorator
@authorization("manager")
def view_salary():
    print("Salary details displayed")
view_salary()
```

# Question : Create an authorization decorator that accepts a single string parameter representing the access level. Allow the function to execute only when the access level is "premium.

```python
def authorization(access):
    def decorator(func):
        def wrapper():
            if access == "premium":
                func()
            else:
                print("Premium Access Required")
        return wrapper
    return decorator
@authorization("premium")
def watch_video():
    print("Video started")
watch_video()
```

# 4. Memoization Decorator (Single Integer Argument) 

# Question: Create a memoization decorator for a function that takes one integer parameter.

```python
def decor(func):
    cache = {}
    def wrapper(n):
        if n in cache:
            return cache[n]
        result = func(n)
        cache[n] = result
        return result
    return wrapper
@decor
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
print(factorial(5))
```
